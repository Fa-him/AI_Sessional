To calculate the regression line, we need to determine "m" the slope of the line, and "c" the constant according to the following equation:

$m = \sum (x - \bar{x})(y - \bar{y}) / \sum{(x - \bar{x})^2} $

$c = \bar{y} - m\bar{x} $

Regression line: $y = mx + c$

```python
meanX = X.mean()
meanY = Y.mean()
```

```python
print(meanX)
print(meanY)
```

```python
n = len(X)
```

```python
print(n)
```

```python
import numpy as np
```

```python
m =(((X-meanX)*(Y-meanY)).sum())/((((X-meanX)**2)).sum())
```

```python
print(m)
```

```python
c= meanY - (m*meanX)
```

```python
print(c)
```

```python
# integer input
new_x = float(input())
```

```python
y = (m * new_x) + c
y
```

```python
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(4,8,100)  #(4-8)---> range
y = (m * x) + c
plt.plot(x,y,color="g",label="Regression Line")
plt.scatter(X,Y, color="r",label="Scatter Plot")
plt.title("Regression")
plt.xlabel("Height (FT)")
plt.ylabel("Weight (KG)")
plt.legend()
plt.show()
```

##Cost Function##
## MSE (Mean Square Error) or L2 Loss ##
$Cost  = \frac{\sum(h(x) - y)^2}{2n}$

```python
# just predicted value
y=(m*X)+c
```

```python
print(y)
```

```python
# y= predicted value, Y= Actual value
#cost = error between actual and predicted value
cost=((y-Y)**2).sum()/(2*n)
#total actual and real y difference
c1=((y-meanY)**2).sum()
print(c1)
```

```python
cost
```

##R2 Coefficient##
R2 is a measure of the goodness of fit of a model. In regression, the R2 coefficient of determination is a statistical measure of how well the regression predictions approximate the real data points.

$R^2  = \frac{\sum(h(x) - \bar{y})^2}{\sum(y - \bar{y})^2}$

```python
#actual Y
R =c1/((Y-meanY)**2).sum()
```

```python
print(R*100)
```

##**Linear Regression Using Scikit-learn**##

```python
data = pd.read_excel('/content/drive/MyDrive/HW data.xlsx')
data.head(7)
```

```python
data.head()
```

```python
#Logistic regression (Regression), KNN (classification), K-means (clustering), Bayes (Classification), Gradient Decent (Optimization), PCA (Feature Selection), DT (Classification)
from sklearn.linear_model import LinearRegression
# metrics
from sklearn.metrics import mean_squared_error
```

```python
data = np.asarray(data)
X = data[:, 0:1]
Y = data[:, -1] # row col range
```

```python
model = LinearRegression()
#knowledge
model = model.fit(X, Y)
print(model.fit(X, Y))
```

```python
prediction = model.predict(X)
print(prediction )
print(model.coef_) # y=mx+c, calculate the c of the model
print(model.intercept_) #y=mx+c, calculate the m of the model
result = model.score(X, Y) #calculate the R2 Coficient of the model
print(result)
```
